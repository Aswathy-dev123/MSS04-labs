```python
import math

# Solve ax² + bx + c = 0
def q_solve(a, b, c):

    # Compute the discriminant
    D = b**2 - 4*a*c

    if D >= 0:
        x1 = (-b + math.sqrt(D)) / (2*a)
        x2 = (-b - math.sqrt(D)) / (2*a)

        return x1, x2

    else:
        print("no real solution")
        return None, None


x1, x2 = q_solve(1, 4, 1)

print("x1 =", x1)
print("x2 =", x2)

x1, x2 = q_solve(1, 1, 1)
```

```python

import math

# Complex solutions allowed
# Solve ax² + bx + c = 0

def q_solve(a, b, c):

    # Compute the discriminant
    D = b**2 - 4*a*c

    if D >= 0:
        sq_D = math.sqrt(D)

    else:
        sq_D = math.sqrt(-D) * 1j

    # Calculate solutions
    x1 = (-b + sq_D) / (2*a)
    x2 = (-b - sq_D) / (2*a)

    return x1, x2


# Test the function
x1, x2 = q_solve(1, 4, 1)

print("x1 =", x1)
print("x2 =", x2)

x1, x2 = q_solve(1, 1, 1)

print("x1 =", x1)
print("x2 =", x2)
```

```python
# solution where a,b,c are complex numbers

import cmath

# Solve ax² + bx + c = 0
def q_solve(a, b, c):

    # Compute the discriminant
    D = b**2 - 4*a*c

    # Calculate square root
    sq_D = cmath.sqrt(D)

    # Calculate solutions
    x1 = (-b + sq_D) / (2*a)
    x2 = (-b - sq_D) / (2*a)

    return x1, x2


# Test 1: Real coefficients
x1, x2 = q_solve(1, 4, 1)


print("x1 =", x1)
print("x2 =", x2)


# Test 2: Complex coefficients
x1, x2 = q_solve(1+1j, 2+3j, 4+2j)


print("x1 =", x1)
print("x2 =", x2)
```

```python
import cmath

def q_solve(a, b, c):

    if a == 0:
        raise ValueError("Coefficient a must not be zero")

    discriminant = b**2 - 4*a*c

    x1 = (-b + cmath.sqrt(discriminant)) / (2*a)
    x2 = (-b - cmath.sqrt(discriminant)) / (2*a)

    return x1, x2
```

```python
print(q_solve(1+1j, 2+0j, 1-1j))

print(q_solve(1j, 2, 1))

print(q_solve(1, 2, 5))
```
